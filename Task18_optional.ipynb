```python
filepath='/content/student dataset.zip'
df=pd.read_csv(filepath)
df
```

```python
print(df.head(10))
print(df.tail(5))
print("Shape:", df.shape)
print(df.dtypes)
```

```python
# Missing values
print(df.isnull().sum())

# Duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```python
print(df.describe())

print("Sleep Hours")
print("Min:", df["Sleep_Hours"].min())
print("Max:", df["Sleep_Hours"].max())
print("Mean:", df["Sleep_Hours"].mean())
print("Median:", df["Sleep_Hours"].median())

print("Study Hours")
print("Min:", df["Study_Hours"].min())
print("Max:", df["Study_Hours"].max())
print("Mean:", df["Study_Hours"].mean())
print("Median:", df["Study_Hours"].median())

print("Attendance")
print("Min:", df["Attendance"].min())
print("Max:", df["Attendance"].max())
print("Mean:", df["Attendance"].mean())
print("Median:", df["Attendance"].median())
```

```python
import matplotlib.pyplot as plt

num_cols = ["Sleep_Hours"]

for col in num_cols:
    plt.figure(figsize=(5,3))
    plt.hist(df[col], bins=20)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cat_cols = ["Student_Type"]

for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col,data=df)
    plt.title(col)
    plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.show()
```

```python
print(df["Student_Type"].head())

encoded_df = pd.get_dummies(df, columns=["Student_Type"], drop_first=True)

print(encoded_df.head())
```
